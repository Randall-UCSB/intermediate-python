### Example 6. Create a Numpy array of evenly spaced values in a given range, using arange(stop_val)

#####  np.arrange()
Creates an array of evenly spaced values within a specified range (similar to Python’s built-in range() but returns a NumPy array).

###### Syntax
```
np.arange([start,] stop[, step], dtype=None)

```

###### Example
```import numpy as np

np.arange(5)      
# Output: array([0, 1, 2, 3, 4])

np.arange(2, 10)  
# Output: array([2, 3, 4, 5, 6, 7, 8, 9])

np.arange(1, 10, 2)  
# Output: array([1, 3, 5, 7, 9])

np.arange(0, 1, 0.2)  
# Output: array([0. , 0.2, 0.4, 0.6, 0.8])


```

````python
# We create a rank 1 ndarray that has sequential integers from 0 to 9
x = np.arange(10)

# We print the ndarray
print()
print('x = ', x)
print()

# We print information about the ndarray
print('x has dimensions:', x.shape)
print('x is an object of type:', type(x))
print('The elements in x are of type:', x.dtype) 
````

### Example 9. Create a Numpy array using linspace(start, stop, n), with stop inclusive.

##### linespace()
- Creates an array of evenly spaced numbers over a specified interval, including both the start and stop values.

###### Syntax

```
np.linspace(start, stop, num=50, endpoint=True, retstep=False)

```

###### Examples
  
```
import numpy as np

np.linspace(0, 1, 5)  
# Output: array([0.  , 0.25, 0.5 , 0.75, 1.  ])

np.linspace(0, 1, 5, retstep=True)  
# Output: (array([0.  , 0.25, 0.5 , 0.75, 1.  ]), 0.25)

```

````python
# We create a rank 1 ndarray that has 10 integers evenly spaced between 0 and 25.
x = np.linspace(0,25,10)

# We print the ndarray
print()
print('x = \n', x)
print()

# We print information about the ndarray
print('x has dimensions:', x.shape)
print('x is an object of type:', type(x))
print('The elements in x are of type:', x.dtype) 
````

### Example 12. Create a Numpy array by calling the reshape() function from the output of arange() function.

##### arrange()
Creates an array with evenly spaced values within a given range.

```
np.arange([start,] stop[, step])

```
###### Example   
```
import numpy as np

```

##### reshape()
Changes the shape of an existing array without changing its data.

###### Syntax  
```
array.reshape(new_shape)
```

###### Example 
```
a = np.arange(6)          # array([0, 1, 2, 3, 4, 5])
a.reshape((2, 3))         # Output: array([[0, 1, 2],
                          #                 [3, 4, 5])

```


#### Before

````python
Y = np.arange(20)
print('Y = \n', Y)
````

#### After

````python
# We create a a rank 1 ndarray with sequential integers from 0 to 19 and
# reshape it to a 4 x 5 array 
Y = np.arange(20).reshape(4, 5)

# We print Y
print()
print('Y = \n', Y)
print()

# We print information about Y
print('Y has dimensions:', Y.shape)
print('Y is an object of type:', type(Y))
print('The elements in Y are of type:', Y.dtype) 
````

### Example 16. Create a Numpy array of "Normal" distributed random numbers, using the numpy.random.normal() function.

##### random.normal()
##### 
- Generates random numbers drawn from a normal (Gaussian) distribution.

###### Syntax
```
numpy.random.normal(loc=0.0, scale=1.0, size=None)
```

###### Example
```
import numpy as np

np.random.normal()  
# Output: single value from standard normal distribution (mean=0, std=1)

np.random.normal(loc=10, scale=2, size=5)  
# Output: array of 5 values with mean=10 and std=2

np.random.normal(0, 1, (2, 3))  
# Output: 2x3 array of values from standard normal distribution

```

````python
# We create a 1000 x 1000 ndarray of random floats drawn from normal (Gaussian) distribution
# with a mean of zero and a standard deviation of 0.1.
X = np.random.normal(0, 0.1, size=(1000,1000))

# We print X
print()
print('X = \n', X)
print()

# We print information about X
print('X has dimensions:', X.shape)
print('X is an object of type:', type(X))
print('The elements in X are of type:', X.dtype)
print('The elements in X have a mean of:', X.mean())
print('The maximum value in X is:', X.max())
print('The minimum value in X is:', X.min())
print('X has', (X < 0).sum(), 'negative numbers')
print('X has', (X > 0).sum(), 'positive numbers')
````
