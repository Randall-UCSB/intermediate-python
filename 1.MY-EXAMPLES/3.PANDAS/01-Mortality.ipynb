### loc and iloc 

- use the loc[ ] (location) or iloc[ ] (integer location) accessor

- If you’ve used slices with your Python code, you know that a slice has a starting value, an ending value, and an optional step value with the values separated by colons.
- If you’re using positions instead of labels in your slices, those positions start from 0 and the ending position isn’t included in the result.
- So, for example, 1:3 is a slice that refers to the second and third values, but not the fourth.
- However, with the loc[] accessor, this works a little differently. When you use slices with DataFrames, the ending position is included.
- So, if you use 1:3 for a slice in the loc accessor, you get the rows where the index value is 1, 2, or 3.

### Loc
- You can select a range of rows, individual rows or "every x rows." 

````python
mortality_data.loc[0:1]
````

````python
mortality_data.loc[0:5:2]
````

````python
mortality_data.loc[[0,5,10],['AgeGroup','DeathRate']] 
````

- The columns start at 0

- What happens when you reference a column that is not there?

### Mean Centering
- The mean center is a way to make analysis of data more reasonable since it reduces the impact of outliers.
    - When doing a regression of salary based on years of experience, mean centering will reduce the 
      impact of salary at zero years of experience.
- All of the mean centered values in a set will add to zero.

##### When you have multiple variables, like grades based on study time and sleep time
You can use mean centering to see 
- What each variable does on its own
- What they do together

For example, 

- “If you study more than average, your score goes up.”
- “If you sleep more than average, your score goes up.”
- “If you do both, there’s a slight twist in how they combine.”


````python
mortality_data.DeathRate.mean()
````

### Add a column for the Mean Centered values of Death Rate

## How to Shape the Data

### Indexes
- Default Index: DataFrames use a default index starting at 0; the last index is one less than the total number of rows.
- Custom Indexing: You can use set_index() to assign a column (or columns) as the index, which can aid operations like plotting.
- Non-Unique Index Check: Use verify_integrity=True in set_index() to ensure index values are unique; otherwise, an error is raised.
- Multi-Indexing: Setting the index with multiple columns (e.g., Year and AgeGroup) creates a multi-index, which allows for hierarchical indexing.
- Index Display: Index columns appear visually at a lower level than data columns in output displays.
- Limitations of Index Columns: Once a column becomes an index, it can’t be used as a parameter in methods unless reset.
- Resetting Index: Use reset_index() to convert index columns back to regular columns.
- inplace Parameter: Most methods, including set_index() and reset_index(), have an inplace option (default is False). It's best practice to leave it False.

### Pivot the Data

##### Why Pivot Data in a DataFrame
Pivoting data in a DataFrame is useful when you want to **reshape** or **reorganize** the data to make it easier to analyze, visualize, or report. Here are key reasons:

##### Rearrange Data for Clarity
- **Convert rows into columns** (or vice versa) to make patterns or comparisons more visible.
- *Example*: Turn a column of years into column headers to see each year side by side.

##### Enable Better Analysis
- Summarize data in a more intuitive format, often used in time series or cross-tab analysis.
- *Example*: Show death rates by year (as columns) and age group (as rows).

##### Support for Plotting
- Many visualization tools work better when data is pivoted into a "wide format" where each column represents a distinct series.

##### Facilitate Calculations
- Pivoted data can make it easier to perform element-wise operations or aggregations across categories.

##### Tidy Data Principles
- Pivoting can help achieve a "tidy" format where:
  - Each variable forms a column.
  - Each observation forms a row.
  - Each type of observational unit forms a table.
